# Comparison of Bigquery and Kaggle Datasets

```python
import numpy as np
import pandas as pd
```

```python
bq_df = pd.read_csv("../data/real_bitcoin_blocks_raw.csv")
kaggle_df = pd.read_csv("../data/kaggle_dataset.csv")
```

```python
bq_df
```

```python
kaggle_df
```

```python
merged = kaggle_df.merge(
    bq_df, left_on="height", right_on="number", suffixes=("_kaggle", "_bq")
)
mismatches = merged[merged["total_fees"] != merged["total_fee_satoshis"]]
print(len(mismatches), "mismatched rows out of", len(merged))
```

```python
outer = kaggle_df.merge(bq_df, left_on="height", right_on="number", how="outer", indicator=True)
print(outer["_merge"].value_counts())
```
