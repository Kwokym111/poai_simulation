```python
from google.cloud import bigquery

client = bigquery.Client(project="poai-research")

query = """
SELECT *
FROM `poai-research.poai_data.block_volumes`
LIMIT 5
"""

df = client.query(query).to_dataframe()
print(df)
print(df.shape)
print(df.dtypes)
```
